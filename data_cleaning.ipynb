```python
import pandas as pd

data = pd.read_csv("credit_card_transactions.csv")

pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

# 1. Drop irrelevant coulumns 
data = data.drop(columns=["Unnamed: 0","unix_time","street"])

# 2. Drop duplicates
data = data.drop_duplicates()

# 3. handle missing data
data = data.fillna({"merch_zipcode": "Unknown"}) 

# 4. merge name 
data["full_name"] = data["first"] + ' ' + data["last"]
#5. drop columns
data = data.drop(columns=["first","last"])
#6. rename columns
data = data.rename(columns= {"trans_date_trans_time": "transaction_time","cc_num": "credit_card_number",
                             "amt": "amount","lat": "latitude","long": "longitude","city_pop": "city_population",
                             "dob": "birth_date","trans_num": "transaction_number","merch_lat": "merchant_latitude",
                             "merch_long": "merchant_longitude","merch_zipcode": "merchant_zip_code"})


dim_customers = data[["credit_card_number","full_name","gender","city",
                      "state","zip","latitude","longitude",
                      "city_population","job","birth_date"]]
# ---------------------------------------------dim_customers-------------------------------------------------
dim_customers = dim_customers.drop_duplicates(subset="credit_card_number")
dim_customers.head()
#print(data.head().T)

# ---------------------------------------------dim_merchants-------------------------------------------------
dim_merchants = data[["merchant","category","merchant_latitude",
                      "merchant_longitude","merchant_zip_code"]]
dim_merchants = dim_merchants.drop_duplicates(subset=["merchant"])
dim_merchants.head()

# ---------------------------------------------fact_transactions-------------------------------------------------
fact_transactions = data[["transaction_number","credit_card_number","merchant",
                          "amount","transaction_time","is_fraud"]]
fact_transactions.head()
```
